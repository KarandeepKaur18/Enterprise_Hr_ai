```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, roc_auc_score

PROCESSED_PATH = "../data/processed"
print("Loading engineered features...")
df = pd.read_csv(f"{PROCESSED_PATH}/attrition_features.csv")

# Separate target and features
# (Ensure 'Attrition_Label' matches what we named it in Step 5!)
y = df['Attrition_Label']
X_raw = df.drop(columns=['Attrition_Label', 'EmployeeNumber'], errors='ignore')

# Tech magic: Fast one-hot encoding for all remaining text columns
X = pd.get_dummies(X_raw, drop_first=True)

print(f"Final feature count for modeling: {X.shape[1]}")

# Train/Test Split (stratified to maintain the Yes/No ratio)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

print("Training Logistic Regression Baseline...")
model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)

# Evaluate using probabilities
probs = model.predict_proba(X_test)[:, 1]
roc_auc = roc_auc_score(y_test, probs)

print("\n--- Baseline Results ---")
print(f"ROC-AUC Score: {roc_auc:.4f}")
print("\nClassification Report:")
print(classification_report(y_test, model.predict(X_test)))
```

```python
import pandas as pd
import os

RAW_PATH = "../data/raw"
PROCESSED_PATH = "../data/processed"

# Tech magic: automatically create the processed folder if it doesn't exist
os.makedirs(PROCESSED_PATH, exist_ok=True)

print("Loading raw data...")
df_attrition = pd.read_csv(f"{RAW_PATH}/employee_attrition.csv")

# 1. Drop absolute duplicates
initial_shape = df_attrition.shape
df_attrition = df_attrition.drop_duplicates()
print(f"Dropped {initial_shape[0] - df_attrition.shape[0]} duplicate rows.")

# 2. Handle missing values (Basic fill for the MVP)
# Filling numerical NaNs with median and categorical with 'Unknown'
for col in df_attrition.columns:
    if df_attrition[col].dtype == 'object':
        df_attrition[col] = df_attrition[col].fillna('Unknown')
    else:
        df_attrition[col] = df_attrition[col].fillna(df_attrition[col].median())

# 3. Save the clean copy
processed_file = f"{PROCESSED_PATH}/employee_attrition_processed.csv"
df_attrition.to_csv(processed_file, index=False)

print(f"Clean data successfully saved to: {processed_file}")
print("New Shape:", df_attrition.shape)
```

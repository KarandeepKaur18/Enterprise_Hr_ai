```python
import pandas as pd

RAW_PATH = "../data/raw"
print("Loading Engagement Data...")

# Load the performance dataset
# Make sure this matches the exact filename in your raw folder!
df_perf = pd.read_csv(f"{RAW_PATH}/Employee_Performance_Dataset.csv")

# Quick check on the column names to ensure we grab the right ones
print("Columns available:", list(df_perf.columns))

# Assuming we have columns like 'Department' and 'EngagementScore'
# Let's write a quick try-except block just in case the column names are slightly different
try:
    print("\n--- Average Performance by Department ---")
    # Grouping by department and calculating the mean score
    dept_engagement = df_perf.groupby('Department')['Performance Score'].mean().sort_values(ascending=False)
    print(dept_engagement)

    print("\n--- Critical Attention: Bottom 5 Lowest Performance Employees ---")
    # Grabbing the lowest scores
    lowest_engagement = df_perf[['Employee ID', 'Department', 'Performance Score']].sort_values(by='Performance Score').head(5)
    print(lowest_engagement)
    
except KeyError as e:
    print(f"\nOops! Looks like your real data uses a different column name: {e}")
```

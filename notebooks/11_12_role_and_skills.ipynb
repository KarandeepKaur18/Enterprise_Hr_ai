```python
import pandas as pd
import random
import os

PROCESSED_PATH = "../data/processed"

print("--- Step 11: Role Intelligence ---")
# Loading the occupation master list just to verify it exists
df_roles = pd.read_csv("../data/raw/occupation_data.csv")
print(f"Total Master Roles Available: {df_roles.shape[0]}")

print("\n--- Step 12: Generating Employee Skills ---")
# Load attrition data to get the real employee IDs
df_attrition = pd.read_csv(f"{PROCESSED_PATH}/employee_attrition_processed.csv")

# Find the ID column (usually EmployeeNumber, but we use a fallback just in case)
id_col = 'EmployeeNumber' if 'EmployeeNumber' in df_attrition.columns else df_attrition.columns[0]

# Our pool of tech skills
tech_skills = ['Python', 'SQL', 'AWS', 'Docker', 'MLOps', 'Machine Learning', 'Data Analysis', 'Java', 'Generative AI']

# Generate 2 to 4 random skills for every single employee
employee_skills_data = []
for emp_id in df_attrition[id_col]:
    emp_skills = random.sample(tech_skills, random.randint(2, 4))
    for skill in emp_skills:
        employee_skills_data.append({'EmployeeID': emp_id, 'CurrentSkill': skill})
        
df_emp_skills = pd.DataFrame(employee_skills_data)

# Save the mock table for the next step
mock_skills_file = f"{PROCESSED_PATH}/employee_skills_mock.csv"
df_emp_skills.to_csv(mock_skills_file, index=False)

print(f"Tech magic success! Generated {len(df_emp_skills)} skill records for {len(df_attrition)} employees.")
print(f"Saved to: {mock_skills_file}")
print("\nPreview:")
print(df_emp_skills.head())
```

```python
import os
import json
import shutil
from datetime import datetime

# Define versions and paths
VERSION = "v1.0"
BASE_MODEL_DIR = "../models"
VERSION_DIR = f"{BASE_MODEL_DIR}/{VERSION}"
OLD_MODEL_PATH = f"{BASE_MODEL_DIR}/attrition_pipeline.joblib"
NEW_MODEL_PATH = f"{VERSION_DIR}/attrition_pipeline.joblib"
METADATA_PATH = f"{VERSION_DIR}/metadata.json"

# Create version folder
os.makedirs(VERSION_DIR, exist_ok=True)

# Move the winning model we created in Step 7 to the versioned folder
if os.path.exists(OLD_MODEL_PATH):
    shutil.move(OLD_MODEL_PATH, NEW_MODEL_PATH)
    print("Model successfully moved to versioned folder.")

# Create the metadata dictionary
metadata = {
    "model_name": "Attrition Prediction Model",
    "version": VERSION,
    "algorithm": "Random Forest",
    "training_date": datetime.now().strftime("%Y-%m-%d"),
    "roc_auc": 0.796, 
    "notes": "Winning baseline model from MVP showdown"
}

# Save metadata to JSON
with open(METADATA_PATH, "w") as f:
    json.dump(metadata, f, indent=4)

print(f"\nEnterprise tech magic applied! Model safely versioned in: {VERSION_DIR}")
print("Generated Metadata:")
print(json.dumps(metadata, indent=4))
```

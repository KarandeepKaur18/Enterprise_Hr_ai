```python
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
 
pd.set_option("display.max_columns", None)
DATA_PATH = "../data/raw"
 
print(os.listdir(DATA_PATH))
```

```python
# Load the main attrition dataset
df = pd.read_csv(f"{DATA_PATH}/employee_attrition.csv")

# 1. Check the shape (rows, columns)
print("Shape:", df.shape)

# 2. Look at the basic info and data types
df.info()

# 3. Check for missing values
print("\nMissing Values:")
print(df.isnull().sum().sort_values(ascending=False).head(10))

# 4. Check for pure duplicates
print("\nDuplicates:", df.duplicated().sum())

# 5. Check the target balance (assuming your column is named 'Attrition' or 'RiskLabel')
# Change 'Attrition' to match your actual column name if it throws an error!
print("\nAttrition Balance (%):")
print(df['Attrition'].value_counts(normalize=True) * 100)

df.head()
```

```python
# Hunting for any column with 'id' in the name across all files
for file in os.listdir(DATA_PATH):
    if file.endswith('.csv'):
        # Just loading the headers to save memory
        temp_df = pd.read_csv(f"{DATA_PATH}/{file}", nrows=0) 
        
        # List comprehension to find 'id' matches
        id_cols = [c for c in temp_df.columns if 'id' in c.lower()]
        
        print(f"{file} -> Possible Keys: {id_cols}")
```

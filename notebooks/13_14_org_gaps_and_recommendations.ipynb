```python
from collections import Counter
import pandas as pd

print("--- Step 14: Organization-Wide Skill Gap ---")

def calculate_org_gaps(all_missing_skills_list):
    """Counts missing skills and assigns severity based on business rules."""
    # Tech magic: Counter does all the heavy lifting instantly
    gap_counts = Counter(all_missing_skills_list)
    
    org_gaps = []
    for skill, count in gap_counts.items():
        # Applying your exact severity rules
        if count >= 100:
            severity = "HIGH"
        elif count >= 50:
            severity = "MEDIUM"
        else:
            severity = "LOW"
            
        org_gaps.append({"Skill": skill, "Missing_Count": count, "Severity": severity})
        
    return pd.DataFrame(org_gaps).sort_values(by="Missing_Count", ascending=False)

# Mocking some data: let's say 120 people miss MLOps, 75 miss Cloud, 30 miss Python
mock_org_gaps = (['MLOps'] * 120) + (['Cloud'] * 75) + (['Generative AI'] * 30)
org_gap_df = calculate_org_gaps(mock_org_gaps)
print(org_gap_df.to_string(index=False))


print("\n--- Step 15: Upskilling Recommendation Engine ---")

def generate_recommendations(missing_skills):
    """Version 1: Simple rules mapping missing skills to training."""
    recommendations = []
    
    # Plain if/else rules as requested in the MVP plan
    for skill in missing_skills:
        if skill == "MLOps":
            recommendations.append("Course: Deploying and Monitoring Machine Learning Systems")
        elif skill == "Cloud" or skill == "AWS":
            recommendations.append("Course: AWS Cloud Practitioner Essentials")
        elif skill == "Python":
            recommendations.append("Course: Advanced Python for Data Science")
        else:
            recommendations.append(f"Search Internal LMS for: {skill}")
            
    return recommendations

# Testing the recommendation engine on Employee 101's gap
employee_101_gap = ['MLOps', 'AWS']
recs = generate_recommendations(employee_101_gap)

print(f"Employee missing: {employee_101_gap}")
print("Recommended Actions:")
for rec in recs:
    print(f" -> {rec}")
```

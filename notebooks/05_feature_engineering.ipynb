```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

PROCESSED_PATH = "../data/processed"
print("Loading clean data...")
df = pd.read_csv(f"{PROCESSED_PATH}/employee_attrition_processed.csv")

print("Executing feature engineering hacks...")

# Hack 1: Experience Ratio (Years at company vs Total working years)
# Using a quick conditional to avoid dividing by zero
if 'TotalWorkingYears' in df.columns and 'YearsAtCompany' in df.columns:
    df['ExperienceRatio'] = df.apply(
        lambda row: row['YearsAtCompany'] / row['TotalWorkingYears'] if row['TotalWorkingYears'] > 0 else 0, 
        axis=1
    )

# Hack 2: Overall Satisfaction Score
# Combining multiple satisfaction metrics into one powerful feature
if 'JobSatisfaction' in df.columns and 'EnvironmentSatisfaction' in df.columns:
    df['OverallSatisfaction'] = (df['JobSatisfaction'] + df['EnvironmentSatisfaction']) / 2

# Hack 3: Label Encoding
# The model needs numbers, not 'Yes' or 'No'
le = LabelEncoder()
if 'Attrition' in df.columns:
    df['Attrition_Label'] = le.fit_transform(df['Attrition'])
    
# Drop the original target text column and any useless static columns
cols_to_drop = ['Attrition', 'EmployeeCount', 'StandardHours', 'Over18']
df_engineered = df.drop(columns=[c for c in cols_to_drop if c in df.columns])

# Save the engineered dataset for our baseline model
engineered_file = f"{PROCESSED_PATH}/attrition_features.csv"
df_engineered.to_csv(engineered_file, index=False)

print("Tech magic complete! Dataset is ready for the ML engine.")
print("New Shape:", df_engineered.shape)
```

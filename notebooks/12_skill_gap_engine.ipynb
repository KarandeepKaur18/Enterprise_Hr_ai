```python
print("--- Step 13: Skill Gap Engine ---")

def calculate_skill_gap(required_skills, employee_skills):
    """
    Tech magic: Uses Python sets to instantly find missing skills.
    Required skills MINUS employee skills.
    """
    req_set = set(required_skills)
    emp_set = set(employee_skills)
    
    # The magical set subtraction
    gap = req_set - emp_set
    return list(gap)

# Testing the logic exactly as outlined in your project notes
target_role_skills = ['Python', 'SQL', 'MLOps', 'Docker', 'AWS']
employee_101_skills = ['Python', 'SQL', 'AWS']

missing_skills = calculate_skill_gap(target_role_skills, employee_101_skills)

print(f"Target Role Requires: {target_role_skills}")
print(f"Employee Currently Has: {employee_101_skills}")
print(f"\nCalculated Skill Gap: {missing_skills}")
```
